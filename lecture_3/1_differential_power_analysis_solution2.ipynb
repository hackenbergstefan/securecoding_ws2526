In this example we want to attack an AES encryption using the _Differential-Power-Analysis_ introduces by Kocher et al. 1999.

The Differential-Power-Analysis has few prerequisites or in other words: It can be applied if the following conditions hold true:
- The device holds a **secret** (e.g. a key)
- The **secret is constant** during all measurements
- External **varying input** can be fed into the device
- The secret "**collides**" with external input in a **known algorithm**.

Notes:
- "Collide" means: The input and the secret are processed together in an algorithm
- Varying input: It is not required that the input can be determined by the attacker. It is sufficient that the attacker knows the input and the input is sufficiently random.
- If the input is unknown but varying the attack also works if the output is known by the attacker.

## 2. Develop an attack

The term _Differential_ in Differential-Power-Analysis is not chosen by accident.
We will use differences to develop an attack that reveals the secret key.

#### Basic principle of a Differential-Power-Analysis

1. Divide the set of traces (with random input) into **two partitions** using a **guess**.
2. Calculate the average of each partition.
3. Calculate the **absolute difference** of the two averages.
4. The maximal difference corresponds to the correct guess.

But what does "partition using a guess" mean?

In this exercise we don't want to reveal the key but understand why differences are good to reveal data.

1. Record 1000 traces using random input for the first input byte, i.e. use `input=lambda _: [random.randint(0, 255)] + 15 * [0]`.
2. Divide the data into two sets `data_msb_0` and `data_msb_1` depending on the msb of the random byte. </br>
   Hints:
   - The first byte of `input` of trace `i` can be accessed using `data[i]["input"][0]`.
   - The MSB of a byte `x` can be calculated using `x & 0x80`

3. Create an numpy array of the result: `data_msb_0 = np.array(data_msb_0)`, `data_msb_1 = np.array(data_msb_1)`
4. Calculate the "mean-trace" of each set: `mean_trace_msb_0 = np.mean(data_msb_0["trace"], axis=0)`, `mean_trace_msb_1 = np.mean(data_msb_1["trace"], axis=0)`.
   Lean back and think about the result. </br>
   Hints:
   - The dimensions of a numpy array can be shown using `data.shape`

5. Plot the absolute difference between the two mean-traces.
   What do you see and why?

   If you can explain, you understood one main principle of DPA! 💪

````python
# Record some traces with random input

def random_bytes(index):
    return random.randbytes(16)

data = capture_device.capture(
    number_of_traces=1000,
    input=random_bytes,
)
````

````python
# Plot and see that all traces look equal

fig = pgo.Figure()
for trace in data[:1]:
    fig.add_trace(pgo.Scatter(y=trace["trace"]))
fig.show()
````

````python
def msb(value):
    return value >> 7


def partition(traces, selection_function):
    """
    Return two sets of traces:
    1. All traces with msb(input[0]) == 1
    2. All traces with msb(input[0]) == 0
    """
    set1 = [trace["trace"] for trace in traces if selection_function(trace) == 0]
    set2 = [trace["trace"] for trace in traces if selection_function(trace) != 0]
    return set1, set2


def partition_and_average(traces, selection_function):
    """Return the average of the two sets generated by partition."""
    set1, set2 = partition(traces, selection_function)
    return np.mean(np.array(set1), axis=0), np.mean(np.array(set2), axis=0)


# Plot the difference between the mean traces
fig = pgo.Figure()
for index in range(16):
    mean1, mean2 = partition_and_average(data, lambda trace: msb(trace["input"][index]))
    fig.add_trace(pgo.Scatter(y=np.abs(mean1 - mean2)))
fig.show()

# => For each input byte we see a significant difference at a specific location
# => The location of the peak corresponds to the timestamp of the processing of the respective input byte
````

````python
from lascar.tools.aes import sbox

secret_key = [
    0x01,
    0x02,
    0x03,
    0x04,
    0x05,
    0x06,
    0x07,
    0x08,
    0x69,
    0x6E,
    0x66,
    0x69,
    0x6E,
    0x65,
    0x6F,
    0x6E,
]

# Assume we know the key:
# Plot the differences looking at the SBOX output
fig = pgo.Figure()
for index in range(16):
    mean1, mean2 = partition_and_average(
        data,
        lambda trace: msb(sbox[trace["input"][index] ^ secret_key[index]]),
    )
    fig.add_trace(pgo.Scatter(y=np.abs(mean1 - mean2)))
fig.show()

# => Again: Peaks where the respective SBOX output is processed
````

````python
# Guess all key bytes are 0xff

# Plot the differences looking at the SBOX output
fig = pgo.Figure()
for index in range(16):
    mean1, mean2 = partition_and_average(
        data,
        lambda trace: msb(sbox[trace["input"][index] ^ 0xff]),
    )
    fig.add_trace(pgo.Scatter(y=np.abs(mean1 - mean2)))
fig.update_layout(yaxis_range=[0, 0.02])
fig.show()

# => No peaks!
````

````python
# Test all key bytes for index 0

# Plot the differences looking at the SBOX output
fig = pgo.Figure()
for guess in range(256):
    mean1, mean2 = partition_and_average(
        data,
        lambda trace: msb(sbox[trace["input"][5] ^ guess]),
    )
    fig.add_trace(pgo.Scatter(y=np.abs(mean1 - mean2), name=f"guess: {guess}"))
fig.update_layout(yaxis_range=[0, 0.02])
fig.show()

# => The correct guess reveals a peak, all other guesses are below
````

````python
def dpa_aes(traces, index):
    maximas = []
    # Iterate over all possible key byte values
    for guess in range(256):
        # Calculate the partitions
        mean1, mean2 = partition_and_average(
            data,
            lambda trace: msb(sbox[trace["input"][index] ^ guess]),
        )
        # Save the current maximum difference
        maximas.append(max(np.abs(mean1 - mean2)))
    # The correct key byte value is the index of the highest maximum
    return int(np.argmax(maximas))

# Iterate over all key byte indices
# to reveal the full key
for key_index in range(16):
    print("key index: ", key_index, "key value: ", hex(dpa_aes(data, key_index)))
````
